```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the final merged dataset from Exercise 2.2
df = pd.read_csv("citibike_weather_2022.csv")
```

```python
# Ensure date is in datetime format
df["date"] = pd.to_datetime(df["date"])
```

```python
# Check if data is loaded correctly
print("Dataset Columns:", df.columns)
print("First 5 Rows:\n", df.head())
```

```python
# Filter only temperature data (using the 'value' column)
weather_df = df[df["datatype"] == "TAVG"]  # Ensure this filters average temperature data
```

```python
# Plot temperature over time
plt.figure(figsize=(12, 5))
plt.plot(weather_df["date"], weather_df["value"], label="Avg Temperature (°C)", color="red")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Trends Over 2022")
plt.legend()
plt.grid()
plt.show()
```

```python
# Convert 'started_at' to datetime format and extract the date
df["trip_date"] = pd.to_datetime(df["started_at"]).dt.date
```

```python
# Count trips per day
trip_counts = df.groupby("trip_date")["ride_id"].count().reset_index()
trip_counts.rename(columns={"ride_id": "trip_count"}, inplace=True)
```

```python
# Convert trip_date to datetime for merging
trip_counts["trip_date"] = pd.to_datetime(trip_counts["trip_date"])
weather_df["date"] = pd.to_datetime(weather_df["date"])

# Merge bike trip counts with weather data
merged_df = trip_counts.merge(weather_df[["date", "value"]], left_on="trip_date", right_on="date", how="left")

# Rename temperature column for clarity
merged_df.rename(columns={"value": "avg_temperature"}, inplace=True)
```

```python
# Save merged dataset for future use
merged_df.to_csv("trip_weather_merged.csv", index=False)
```

```python
# Check result
print("Merged Dataset Columns:", merged_df.columns)
print("First 5 Rows:\n", merged_df.head())
```

```python
fig, ax1 = plt.subplots(figsize=(12, 5))
```

```python
# Plot bike trip counts on primary axis
ax1.plot(merged_df["trip_date"], merged_df["trip_count"], label="Bike Trips", color="blue")
ax1.set_xlabel("Date")
ax1.set_ylabel("Trip Count", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")
```

```python
# Create secondary y-axis for temperature
ax2 = ax1.twinx()
ax2.plot(merged_df["trip_date"], merged_df["avg_temperature"], label="Avg Temperature (°C)", color="red")
ax2.set_ylabel("Temperature (°C)", color="red")
ax2.tick_params(axis="y", labelcolor="red")
```

```python
# Title & Grid
plt.title("Bike Trip Counts vs Temperature in 2022")
fig.tight_layout()
plt.grid()
plt.show()
```

## Analysis of Bike Trips & Temperature Trends (Exercise 2.3)

### Steps Taken:
1. **Loaded the cleaned dataset** from Exercise 2.2.
2. **Filtered temperature data** using the 'value' column and plotted it over time.
3. **Counted bike trips per day** and merged it with temperature data.
4. **Created a dual-axis plot** to show the relationship between bike trips and temperature.

### Why Matplotlib?
- Used `plt.plot()` for simple line plots.
- Used `twinx()` for a secondary y-axis to compare two variables.
- Ensured clear labeling and legends for easy interpretation.

**Observations:**
- Higher bike trip counts are observed in warmer months.
- Fewer trips happen in extreme cold or high heat.
